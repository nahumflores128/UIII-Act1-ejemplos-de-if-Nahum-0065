```python
# 1. Verificación de números (Uso de if, elif, else y operadores relacionales)
# Este ejemplo determina si un número ingresado es positivo, negativo o cero utilizando comparaciones directas
numero = 5 # El número se ha establecido directamente en el código.

if numero > 0:                             # Evalúa si el número es mayor a cero (relacional) [5].
    print("El número es positivo.")        # Se ejecuta si la condición anterior es verdadera [5].
elif numero < 0:                           # Evalúa si es menor a cero si la primera falló [5].
    print("El número es negativo.")        # Se ejecuta si es negativo [5].
else:                                      # Se ejecuta si ninguna de las anteriores es verdadera [5].
    print("El número es cero.")            # Resultado final si el número es exactamente 0 [5].
```

```python
# 2. Identificación de años bisiestos (Uso de operadores aritméticos y lógicos)
# Para saber si un año es bisiesto, se utiliza el operador de módulo (%) para verificar la divisibilidad y los operadores lógicos and y or para combinar las reglas

año = 2024 # El año se ha establecido directamente en el código.

# Un año es bisiesto si es divisible por 4 y NO por 100, O si es divisible por 400 [6].
if (año % 4 == 0 and año % 100 != 0) or (año % 400 == 0):
    print(f"{año} es un año bisiesto.") # Se ejecuta si la lógica compuesta es verdadera [6].
else:
    print(f"{año} no es un año bisiesto.") # Se ejecuta si la lógica compuesta es falsa [6].
```

```python
# 3. Sistema de descuentos (Uso de if-elif-else y operadores relacionales)
# Este código calcula un descuento basado en la cantidad de artículos, demostrando cómo manejar múltiples rangos de valores
cantidad = int(input("¿Cuántos artículos compraste? ")) # Captura la cantidad de artículos [5].

if cantidad >= 10:                      # Comprueba si la cantidad es 10 o más [5].
    descuento = 0.2                     # Asigna 20% de descuento [5].
elif cantidad >= 5:                     # Si no es >= 10, comprueba si es 5 o más [5].
    descuento = 0.1                     # Asigna 10% de descuento [5].
else:                                   # Si es menor a 5 [5].
    descuento = 0                       # No aplica descuento [5].

print(f"Tu descuento es del {descuento * 100}%.") # Muestra el resultado final calculado [6].
```

```python
# 4. Verificación de rango (Uso de operadores lógicos)
# Se puede verificar si un valor se encuentra dentro de un intervalo específico combinando condiciones con and o usando la sintaxis abreviada de Python
numero = int(input("Ingresa un número: ")) # Obtiene el número del usuario [8].

if 10 <= numero <= 20:                     # Verifica si el número está entre 10 y 20 inclusive [8].
    print(f"El número {numero} está entre 10 y 20.") # Confirmación del rango [8].
else:
    print(f"El número {numero} no está entre 10 y 20.") # Aviso de fuera de rango [8].
```

```python
# 5. Determinación de paridad (Uso de operador aritmético de módulo)
# El operador % es fundamental para determinar si un número es par o impar al dividirlo por 2
num = int(input("Ingresa un entero: ")) # Lee un número entero [9].

if num % 2 == 0:                        # Si el residuo de dividir por 2 es cero [7, 9].
    print("El número es par.")          # Mensaje para números pares [9].
else:                                   # Si el residuo es distinto de cero (usualmente 1) [9].
    print("El número es impar.")        # Mensaje para números impares [9].
```
